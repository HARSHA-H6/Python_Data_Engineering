```python
import numpy as np
import pandas as pd

data = pd.read_csv("Data\\ipl_2025.csv")
# Shape of dataset
print("Shape:", data.shape)
# First 5 rows
print("First 5 rows:\n", data[:5])
```

```python
data = data.reset_index()
data["win_by_runs"] = 0
```

```python
data.head()
```

```python
data["win_by_runs"] = 0
data["win_by_runs"] = ((data['first_ings_score'] - data['second_ings_score']).clip(lower=0).where((data['match_result'] == "completed"))
)
data
```

```python
data["win_by_wickets"] = (10 - data["second_ings_wkts"].where(data["win_by_runs"] == 0,11))
data
```

```python
data.shape
```

```python
# Assuming win_by_runs is column index 11 and win_by_wickets is index 12 (adjust as per your dataset)
win_by_runs = data['win_by_runs'].astype(float)
win_by_wickets = data['win_by_wickets'].astype(float)

print("Max runs:", np.max(win_by_runs))
print("Min runs:", np.min(win_by_runs))

print("Mean wickets:", np.mean(win_by_wickets))
print("Median wickets:", np.median(win_by_wickets))
print("Std Dev wickets:", np.std(win_by_wickets))
```

```python
# Matches where winning team scored more than 50 runs
runs_above_50 = data[win_by_runs > 50]

# Matches decided by wickets (runs = 0)
decided_by_wickets = data[win_by_runs == 0]
```
